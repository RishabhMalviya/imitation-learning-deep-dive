```python
import os
os.environ.setdefault("MUJOCO_GL", "osmesa")

import metaworld
import numpy as np
import gymnasium as gym


print(gym.__version__)
```

```python
env = gym.make(
    'Meta-World/MT1', env_name='pick-place-v3',
    seed=42,
    render_mode='rgb_array', camera_name='corner2'
)
obs, _ = env.reset()
```

```python
print("--- Pick-Place-V3 Observation Spaces ---")
print(f"End Effector XYZ Coordinates: {obs[:3]}")
print(f"End Effector Opening Level: {obs[3:4]}")

print(f"Object 1 XYZ Coordinates: {obs[4:7]}")
print(f"Object 1 Quaternion: {obs[7:11]}")

print(f"Object 1 XYZ Coordinates: {obs[11:14]}")
print(f"Object 1 Quaternion: {obs[14:18]}")

print(f"Previous State: {obs[18:36]}")

print(f"Goal XYZ Coordinates (Visible): {obs[36:39]}")
```
